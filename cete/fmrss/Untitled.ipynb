```python
from engines.DataLoader import DataLoader

from engines.FiveMinuteMarketDataAligner import (
    FiveMinuteMarketDataAligner
)


SYMBOLS_FILE = (
    "/home/devinderjeet/fmrss/config/shares.txt"
)

DATA_FOLDER = (
    "/home/hadoop/shareMarket_Data"
)

NIFTY_FILENAME = (
    "NIFTY%2050_5mins.txt"
)


# =========================================================
# Step 1: Load Shares and NIFTY
# =========================================================

data_loader = DataLoader(
    symbols_file=SYMBOLS_FILE,
    data_folder=DATA_FOLDER,
    nifty_filename=NIFTY_FILENAME,
    candle_limit=1500
)

shares_data, nifty_data = (
    data_loader.load_market_data()
)


# =========================================================
# Verify NIFTY Exclusion
# =========================================================

nifty_aliases = {
    "NIFTY",
    "NIFTY50",
    "NIFTY 50",
    "NIFTY_50",
    "NIFTY%2050"
}

incorrect_symbols = (
    set(shares_data.keys())
    & nifty_aliases
)

if incorrect_symbols:

    raise RuntimeError(
        f"NIFTY was incorrectly loaded as a share: "
        f"{incorrect_symbols}"
    )

print(
    f"Tradable shares loaded : "
    f"{len(shares_data)}"
)

print(
    f"NIFTY candles loaded   : "
    f"{len(nifty_data)}"
)


# =========================================================
# Step 2: Align Shares with NIFTY
# =========================================================

aligner = FiveMinuteMarketDataAligner(
    minimum_aligned_candles=450,
    minimum_alignment_ratio=0.90
)

aligned_data, alignment_report = (
    aligner.align_all(
        shares_data=shares_data,
        nifty_data=nifty_data
    )
)


# =========================================================
# Final Verification
# =========================================================

if "NIFTY%2050" in aligned_data:

    raise RuntimeError(
        "NIFTY%2050 must not be present "
        "inside aligned_data"
    )

print(
    f"Successfully aligned shares: "
    f"{len(aligned_data)}"
)


# =========================================================
# Inspect Alignment Report
# =========================================================

display(
    alignment_report[
        [
            "Symbol",
            "Status",
            "Rejection_Reason",
            "Matched_Candles",
            "Alignment_Ratio"
        ]
    ]
)


# =========================================================
# Inspect HINDUNILVR
# =========================================================

if "HINDUNILVR" in aligned_data:

    display(
        aligned_data[
            "HINDUNILVR"
        ].tail(10)
    )
```

```python
import gc
import os
from time import perf_counter

from engines.FiveMinuteMathematicalScoreEngine import (
    FiveMinuteMathematicalScoreEngine
)


# =========================================================
# Configuration
# =========================================================

SCORE_FILE = (
    "/home/devinderjeet/fmrss/report/"
    "MathematicalScoreEngine.xlsx"
)


# =========================================================
# Validate Volatility Data
# =========================================================

if (
    "volatility_data" not in globals()
    or not volatility_data
):
    raise RuntimeError(
        "volatility_data is missing or empty.\n"
        "Run FiveMinuteVolatilityEngine first."
    )


input_symbol_count = len(
    volatility_data
)


# =========================================================
# Create Mathematical Score Engine
# =========================================================

score_engine = FiveMinuteMathematicalScoreEngine(
    # -----------------------------------------------------
    # Mathematical score weights
    # -----------------------------------------------------

    residual_weight=0.35,
    regime_weight=0.25,
    volume_weight=0.20,
    consistency_weight=0.20,

    # -----------------------------------------------------
    # Component normalization references
    # -----------------------------------------------------

    residual_zscore_reference=3.0,
    volume_factor_reference=2.0,

    # -----------------------------------------------------
    # Directional-consistency windows
    # -----------------------------------------------------

    momentum_consistency_window=5,
    reversion_consistency_window=3,

    # -----------------------------------------------------
    # Acceptance threshold
    # -----------------------------------------------------

    minimum_mathematical_score=65.0,

    # -----------------------------------------------------
    # Memory and reporting
    # -----------------------------------------------------

    memory_optimized=True,

    report_output_file=SCORE_FILE,

    save_excel=True,

    # Hide the 80 individual symbol messages.
    print_symbol_details=False,
)


# =========================================================
# Verify Weights
# =========================================================

total_weight = (
    score_engine.residual_weight
    + score_engine.regime_weight
    + score_engine.volume_weight
    + score_engine.consistency_weight
)


if abs(total_weight - 1.0) > 1e-9:
    raise RuntimeError(
        f"Mathematical weights must total 1.0. "
        f"Current total: {total_weight}"
    )


# =========================================================
# Calculate Mathematical Scores
# =========================================================

start_time = perf_counter()


score_data, score_report = (
    score_engine.calculate_all(
        volatility_data=volatility_data
    )
)

# Select only currently accepted shares.
accepted_shares = (
    score_report.loc[
        score_report[
            "Latest_Setup_Accepted"
        ].fillna(False),
        [
            "Symbol",
            "Latest_Stable_Regime",
            "Latest_Residual_Score",
            "Latest_Regime_Score",
            "Latest_Volume_Score",
            "Latest_Consistency_Score",
            "Latest_Mathematical_Score",
            "Latest_Rejection_Reason",
        ],
    ]
    .sort_values(
        "Latest_Mathematical_Score",
        ascending=False,
    )
    .reset_index(drop=True)
)


if accepted_shares.empty:

    print(
        "No shares currently have an "
        "acceptable mathematical setup."
    )

else:

    print(
        f"Acceptable shares: "
        f"{len(accepted_shares)}"
    )

    print(
        accepted_shares.to_string(
            index=False
        )
    )

score_runtime = (
    perf_counter() - start_time
)


# =========================================================
# Validate Results
# =========================================================

if not score_data:
    raise RuntimeError(
        "Mathematical Score Engine returned no data.\n"
        "volatility_data was retained for debugging."
    )


if not os.path.isfile(SCORE_FILE):
    raise RuntimeError(
        f"Mathematical score workbook was not created: "
        f"{SCORE_FILE}\n"
        "volatility_data was retained for debugging."
    )


calculated_symbol_count = len(
    score_data
)


# =========================================================
# Clear Volatility Engine Memory
# =========================================================

removed_variables = []


for variable_name in [
    "volatility_data",
    "volatility_report",
]:

    if variable_name in globals():

        del globals()[variable_name]

        removed_variables.append(
            variable_name
        )


collected_objects = gc.collect()


# =========================================================
# Display Final Summary
# =========================================================

accepted_count = int(
    score_report[
        "Latest_Setup_Accepted"
    ]
    .fillna(False)
    .sum()
)


print(
    "FMRSS : MATHEMATICAL SCORE ENGINE"
)

print(
    f"Successfully calculated : "
    f"{calculated_symbol_count}/"
    f"{input_symbol_count}"
)

print(
    f"Accepted mathematical setups: "
    f"{accepted_count}"
)

print(
    f"Score Engine Runtime: "
    f"{score_runtime:.4f} seconds"
)

print(
    f"Symbols available for Cross-Sectional Ranker: "
    f"{calculated_symbol_count}"
)

print(
    f"Removed from memory: "
    f"{removed_variables}"
)

print(
    f"Garbage-collected objects: "
    f"{collected_objects}"
)

print(
    f"Score workbook: "
    f"{SCORE_FILE}"
)
```

```python
import gc
import io
import importlib
import os
import pandas as pd

from contextlib import redirect_stdout

import engines.FiveMinuteCrossSectionalRanker as ranker_module


# =========================================================
# Configuration
# =========================================================

MATHEMATICAL_SCORE_FILE = (
    "/home/devinderjeet/fmrss/report/"
    "MathematicalScoreEngine.xlsx"
)

RANKING_FILE = (
    "/home/devinderjeet/fmrss/report/"
    "RankingShares.xlsx"
)


# =========================================================
# Validate Mathematical Score Workbook
# =========================================================

if not os.path.isfile(
    MATHEMATICAL_SCORE_FILE
):

    raise FileNotFoundError(
        f"Mathematical Score workbook was not found:\n"
        f"{MATHEMATICAL_SCORE_FILE}"
    )


with pd.ExcelFile(
    MATHEMATICAL_SCORE_FILE,
    engine="openpyxl",
) as workbook:

    if "Accepted_Setups" not in workbook.sheet_names:

        raise RuntimeError(
            "Accepted_Setups worksheet is missing from "
            "MathematicalScoreEngine.xlsx"
        )


# =========================================================
# Reload Ranker
# =========================================================

importlib.invalidate_caches()
importlib.reload(ranker_module)

FiveMinuteCrossSectionalRanker = (
    ranker_module.FiveMinuteCrossSectionalRanker
)


# =========================================================
# Create Ranker
# =========================================================

ranker = FiveMinuteCrossSectionalRanker(
    maximum_selected_symbols=5,

    mathematical_weight=0.60,
    residual_weight=0.25,
    volume_weight=0.15,

    # Do not rank signals older than 15 minutes.
    maximum_staleness_minutes=15,

    mathematical_score_file=(
        MATHEMATICAL_SCORE_FILE
    ),

    accepted_sheet_name=(
        "Accepted_Setups"
    ),

    ranking_output_file=(
        RANKING_FILE
    ),

    save_excel=True,
    memory_optimized=True,
    print_symbol_details=False,
)


# =========================================================
# Rank Accepted Setups
# =========================================================

suppressed_output = io.StringIO()


try:

    with redirect_stdout(
        suppressed_output
    ):

        ranked_data, ranking_report = (
            ranker.rank_from_excel()
        )

finally:

    suppressed_output.close()


# =========================================================
# Validate Results
# =========================================================

if ranking_report is None:

    raise RuntimeError(
        "Cross-Sectional Ranker returned None."
    )


if not os.path.isfile(
    RANKING_FILE
):

    raise RuntimeError(
        f"Ranking workbook was not created:\n"
        f"{RANKING_FILE}"
    )


required_sheets = {
    "Summary",
    "All_Shares",
    "Selected_Shares",
    "Reason_Summary",
}


with pd.ExcelFile(
    RANKING_FILE,
    engine="openpyxl",
) as workbook:

    available_sheets = set(
        workbook.sheet_names
    )


missing_sheets = (
    required_sheets
    - available_sheets
)


if missing_sheets:

    raise RuntimeError(
        f"Ranking workbook is missing worksheets: "
        f"{sorted(missing_sheets)}"
    )


# =========================================================
# Calculate Summary
# =========================================================

accepted_symbol_count = len(
    ranking_report
)


eligible_symbol_count = int(
    ranking_report[
        "Latest_Eligible"
    ]
    .fillna(False)
    .sum()
) if not ranking_report.empty else 0


selected_symbol_count = int(
    ranking_report[
        "Latest_Selected"
    ]
    .fillna(False)
    .sum()
) if not ranking_report.empty else 0


outside_top_n_count = int(
    ranking_report[
        "Latest_Selection_Reason"
    ]
    .eq("OUTSIDE_TOP_N")
    .sum()
) if not ranking_report.empty else 0


stale_symbol_count = int(
    ranking_report[
        "Latest_Selection_Reason"
    ]
    .eq("STALE_INPUT")
    .sum()
) if not ranking_report.empty else 0


ranking_workbook = (
    ranking_report.attrs.get(
        "Excel_File"
    )
    or RANKING_FILE
)


# =========================================================
# Clear Mathematical Score Memory
# =========================================================

removed_variables = []


for variable_name in [
    "score_data",
    "score_report",
    "accepted_shares",
    "score_engine",
]:

    if variable_name in globals():

        del globals()[variable_name]

        removed_variables.append(
            variable_name
        )


# ranked_data is unnecessary because the next engine
# reads RankingShares.xlsx.
if "ranked_data" in globals():

    del ranked_data

    removed_variables.append(
        "ranked_data"
    )


collected_objects = gc.collect()


# =========================================================
# Display Summary
# =========================================================

print("=" * 120)

print(
    f"Accepted setups processed : "
    f"{accepted_symbol_count}"
)

print(
    f"Eligible ranking candidates : "
    f"{eligible_symbol_count}"
)

print(
    f"Outside top five : "
    f"{outside_top_n_count}"
)

print(
    f"Stale candidates : "
    f"{stale_symbol_count}"
)

print(
    f"Latest selected symbols : "
    f"{selected_symbol_count}"
)

print(
    f"Ranking workbook : "
    f"{ranking_workbook}"
)

print(
    f"Removed from memory : "
    f"{removed_variables}"
)

print(
    f"Garbage-collected objects : "
    f"{collected_objects}"
)

print("=" * 120)
```

```python
import gc
import importlib
import os
import pandas as pd

import engines.FiveMinuteRiskEngine as risk_module


# =========================================================
# Reload Risk Engine
# =========================================================

importlib.invalidate_caches()
importlib.reload(risk_module)

FiveMinuteRiskEngine = (
    risk_module.FiveMinuteRiskEngine
)


# =========================================================
# File Locations
# =========================================================

SIGNALS_FILE = (
    "/home/devinderjeet/fmrss/report/"
    "MathematicalSignals.xlsx"
)

TRADING_LEDGER_FILE = (
    "/home/devinderjeet/fmrss/report/"
    "TradingLedger.xlsx"
)

DATA_FOLDER = (
    "/home/hadoop/shareMarket_Data"
)


# =========================================================
# Validate Signal Workbook
# =========================================================

if not os.path.isfile(SIGNALS_FILE):

    raise FileNotFoundError(
        f"Signal workbook not found: "
        f"{SIGNALS_FILE}"
    )


# =========================================================
# Create Risk Engine
# =========================================================

risk_engine = FiveMinuteRiskEngine(
    signals_file=SIGNALS_FILE,
    data_folder=DATA_FOLDER,
    output_file=TRADING_LEDGER_FILE,

    # Read only the five selected signal rows.
    signals_sheet="Selected_Signals",

    initial_capital=100000.0,

    # Maximum account risk per trade = 0.5%.
    risk_per_trade=0.005,

    # Maximum 20% capital allocation per share.
    maximum_allocation_per_trade=0.20,

    # Maximum combined exposure = 80% of capital.
    maximum_portfolio_exposure=0.80,

    maximum_open_positions=5,

    # Reject signals older than one 5-minute candle.
    maximum_signal_age_minutes=5.0,

    atr_window=14,

    # Read only the last 80 rows from each share file.
    atr_history_rows=80,

    mean_reversion_atr_multiplier=1.50,
    mean_reversion_reward_risk=1.50,

    momentum_atr_multiplier=1.20,
    momentum_reward_risk=2.00,

    minimum_stop_percent=0.003,
    maximum_stop_percent=0.03,

    minimum_price=1.0,

    # Prevent unlimited historical ledger growth.
    maximum_ledger_rows=5000,

    memory_optimized=True,

    # Risk decisions will be read later from TradingLedger.xlsx.
    retain_results_in_memory=False,

    save_excel=True,

    # Do not print individual shares.
    print_details=False,
)


# =========================================================
# Calculate Risk Decisions
# =========================================================

risk_data, risk_report = (
    risk_engine.calculate_all()
)


# =========================================================
# Verify TradingLedger.xlsx
# =========================================================

if not os.path.isfile(TRADING_LEDGER_FILE):

    raise RuntimeError(
        f"Trading ledger was not created: "
        f"{TRADING_LEDGER_FILE}"
    )


required_sheets = {
    "Summary",
    "All_Risk_Decisions",
    "Approved_Trades",
    "Rejected_Trades",
    "Trade_Plans",
}


with pd.ExcelFile(
    TRADING_LEDGER_FILE,
    engine="openpyxl",
) as workbook:

    available_sheets = set(
        workbook.sheet_names
    )


missing_sheets = (
    required_sheets - available_sheets
)


if missing_sheets:

    raise RuntimeError(
        f"Trading ledger is missing worksheets: "
        f"{sorted(missing_sheets)}"
    )


approved_trades = int(
    risk_report["Trade_Approved"]
    .fillna(False)
    .sum()
)

rejected_trades = (
    len(risk_report) - approved_trades
)


print("=" * 100)

print(
    f"Risk decisions calculated : "
    f"{len(risk_report)}"
)

print(
    f"Approved trades : "
    f"{approved_trades}"
)

print(
    f"Rejected trades : "
    f"{rejected_trades}"
)

print(
    f"Trading ledger : "
    f"{TRADING_LEDGER_FILE}"
)

print("=" * 100)
```
