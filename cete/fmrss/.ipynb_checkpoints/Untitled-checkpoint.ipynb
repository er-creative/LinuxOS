```python
from engines.DataLoader import DataLoader

from engines.FiveMinuteMarketDataAligner import (
    FiveMinuteMarketDataAligner
)


SYMBOLS_FILE = (
    "/home/devinderjeet/fmrss/config/shares.txt"
)

DATA_FOLDER = (
    "/home/hadoop/shareMarket_Data"
)

NIFTY_FILENAME = (
    "NIFTY%2050_5mins.txt"
)


# =========================================================
# Step 1: Load Shares and NIFTY
# =========================================================

data_loader = DataLoader(
    symbols_file=SYMBOLS_FILE,
    data_folder=DATA_FOLDER,
    nifty_filename=NIFTY_FILENAME,
    candle_limit=1500
)

shares_data, nifty_data = (
    data_loader.load_market_data()
)


# =========================================================
# Verify NIFTY Exclusion
# =========================================================

nifty_aliases = {
    "NIFTY",
    "NIFTY50",
    "NIFTY 50",
    "NIFTY_50",
    "NIFTY%2050"
}

incorrect_symbols = (
    set(shares_data.keys())
    & nifty_aliases
)

if incorrect_symbols:

    raise RuntimeError(
        f"NIFTY was incorrectly loaded as a share: "
        f"{incorrect_symbols}"
    )

print(
    f"Tradable shares loaded : "
    f"{len(shares_data)}"
)

print(
    f"NIFTY candles loaded   : "
    f"{len(nifty_data)}"
)


# =========================================================
# Step 2: Align Shares with NIFTY
# =========================================================

aligner = FiveMinuteMarketDataAligner(
    minimum_aligned_candles=450,
    minimum_alignment_ratio=0.90
)

aligned_data, alignment_report = (
    aligner.align_all(
        shares_data=shares_data,
        nifty_data=nifty_data
    )
)


# =========================================================
# Final Verification
# =========================================================

if "NIFTY%2050" in aligned_data:

    raise RuntimeError(
        "NIFTY%2050 must not be present "
        "inside aligned_data"
    )

print(
    f"Successfully aligned shares: "
    f"{len(aligned_data)}"
)


# =========================================================
# Inspect Alignment Report
# =========================================================

display(
    alignment_report[
        [
            "Symbol",
            "Status",
            "Rejection_Reason",
            "Matched_Candles",
            "Alignment_Ratio"
        ]
    ]
)


# =========================================================
# Inspect HINDUNILVR
# =========================================================

if "HINDUNILVR" in aligned_data:

    display(
        aligned_data[
            "HINDUNILVR"
        ].tail(10)
    )
```

```python
import gc
from time import perf_counter

from engines.FiveMinuteResidualEngine import (
    FiveMinuteResidualEngine
)


# =========================================================
# Validate aligned input
# =========================================================

if (
    "aligned_data" not in globals()
    or not aligned_data
):
    raise RuntimeError(
        "aligned_data is missing or empty.\n"
        "Run FiveMinuteMarketDataAligner first."
    )


# =========================================================
# Create Residual Engine
# =========================================================

residual_engine = FiveMinuteResidualEngine(
    beta_window=375,
    beta_min_periods=350,
    impulse_window=3,

    zscore_window=60,
    zscore_min_periods=50,

    beta_minimum=-1.0,
    beta_maximum=3.0,

    variance_floor=1e-12,
    mad_floor=1e-12,
    zscore_clip=5.0,

    use_cython=True,
    memory_optimized=True,

    # Keep residual_data for Regime Engine.
    retain_results_in_memory=True,

    report_output_file=(
        "/home/devinderjeet/fmrss/report/"
        "ResultResidualEngine.xlsx"
    ),

    # No Excel workbook is created.
    save_excel=False,

    print_symbol_details=False,
)


# =========================================================
# Verify Cython backend
# =========================================================

if residual_engine.robust_stats_backend != "CYTHON":
    raise RuntimeError(
        "Residual Engine selected the PYTHON backend.\n"
        "Compile the Cython extension and restart Jupyter."
    )


# =========================================================
# Calculate Residual Features
# =========================================================

start_time = perf_counter()


residual_data, residual_report = (
    residual_engine.calculate_all(
        aligned_data=aligned_data
    )
)


residual_runtime = (
    perf_counter() - start_time
)


# =========================================================
# Validate Residual Results
# =========================================================

if not residual_data:
    raise RuntimeError(
        "Residual Engine returned no calculated data.\n"
        "aligned_data was retained for debugging."
    )


# =========================================================
# Clear Aligned Data
# =========================================================

if "aligned_data" in globals():
    del globals()["aligned_data"]


collected_objects = gc.collect()


# =========================================================
# Final Information
# =========================================================

print(
    f"Residual runtime: "
    f"{residual_runtime:.2f} seconds"
)

print(
    f"Symbols available for Regime Engine: "
    f"{len(residual_data)}"
)

print(
    "aligned_data removed from memory. "
    f"Collected objects: {collected_objects}"
)
```

```python
import gc
import os
from time import perf_counter

from engines.FiveMinuteMathematicalScoreEngine import (
    FiveMinuteMathematicalScoreEngine
)


# =========================================================
# Configuration
# =========================================================

SCORE_FILE = (
    "/home/devinderjeet/fmrss/report/"
    "MathematicalScoreEngine.xlsx"
)


# =========================================================
# Validate Volatility Data
# =========================================================

if (
    "volatility_data" not in globals()
    or not volatility_data
):
    raise RuntimeError(
        "volatility_data is missing or empty.\n"
        "Run FiveMinuteVolatilityEngine first."
    )


input_symbol_count = len(
    volatility_data
)


# =========================================================
# Create Mathematical Score Engine
# =========================================================

score_engine = FiveMinuteMathematicalScoreEngine(
    # -----------------------------------------------------
    # Mathematical score weights
    # -----------------------------------------------------

    residual_weight=0.35,
    regime_weight=0.25,
    volume_weight=0.20,
    consistency_weight=0.20,

    # -----------------------------------------------------
    # Component normalization references
    # -----------------------------------------------------

    residual_zscore_reference=3.0,
    volume_factor_reference=2.0,

    # -----------------------------------------------------
    # Directional-consistency windows
    # -----------------------------------------------------

    momentum_consistency_window=5,
    reversion_consistency_window=3,

    # -----------------------------------------------------
    # Acceptance threshold
    # -----------------------------------------------------

    minimum_mathematical_score=65.0,

    # -----------------------------------------------------
    # Memory and reporting
    # -----------------------------------------------------

    memory_optimized=True,

    report_output_file=SCORE_FILE,

    save_excel=True,

    # Hide the 80 individual symbol messages.
    print_symbol_details=False,
)


# =========================================================
# Verify Weights
# =========================================================

total_weight = (
    score_engine.residual_weight
    + score_engine.regime_weight
    + score_engine.volume_weight
    + score_engine.consistency_weight
)


if abs(total_weight - 1.0) > 1e-9:
    raise RuntimeError(
        f"Mathematical weights must total 1.0. "
        f"Current total: {total_weight}"
    )


# =========================================================
# Calculate Mathematical Scores
# =========================================================

start_time = perf_counter()


score_data, score_report = (
    score_engine.calculate_all(
        volatility_data=volatility_data
    )
)

# Select only currently accepted shares.
accepted_shares = (
    score_report.loc[
        score_report[
            "Latest_Setup_Accepted"
        ].fillna(False),
        [
            "Symbol",
            "Latest_Stable_Regime",
            "Latest_Residual_Score",
            "Latest_Regime_Score",
            "Latest_Volume_Score",
            "Latest_Consistency_Score",
            "Latest_Mathematical_Score",
            "Latest_Rejection_Reason",
        ],
    ]
    .sort_values(
        "Latest_Mathematical_Score",
        ascending=False,
    )
    .reset_index(drop=True)
)


if accepted_shares.empty:

    print(
        "No shares currently have an "
        "acceptable mathematical setup."
    )

else:

    print(
        f"Acceptable shares: "
        f"{len(accepted_shares)}"
    )

    print(
        accepted_shares.to_string(
            index=False
        )
    )

score_runtime = (
    perf_counter() - start_time
)


# =========================================================
# Validate Results
# =========================================================

if not score_data:
    raise RuntimeError(
        "Mathematical Score Engine returned no data.\n"
        "volatility_data was retained for debugging."
    )


if not os.path.isfile(SCORE_FILE):
    raise RuntimeError(
        f"Mathematical score workbook was not created: "
        f"{SCORE_FILE}\n"
        "volatility_data was retained for debugging."
    )


calculated_symbol_count = len(
    score_data
)


# =========================================================
# Clear Volatility Engine Memory
# =========================================================

removed_variables = []


for variable_name in [
    "volatility_data",
    "volatility_report",
]:

    if variable_name in globals():

        del globals()[variable_name]

        removed_variables.append(
            variable_name
        )


collected_objects = gc.collect()


# =========================================================
# Display Final Summary
# =========================================================

accepted_count = int(
    score_report[
        "Latest_Setup_Accepted"
    ]
    .fillna(False)
    .sum()
)


print(
    "FMRSS : MATHEMATICAL SCORE ENGINE"
)

print(
    f"Successfully calculated : "
    f"{calculated_symbol_count}/"
    f"{input_symbol_count}"
)

print(
    f"Accepted mathematical setups: "
    f"{accepted_count}"
)

print(
    f"Score Engine Runtime: "
    f"{score_runtime:.4f} seconds"
)

print(
    f"Symbols available for Cross-Sectional Ranker: "
    f"{calculated_symbol_count}"
)

print(
    f"Removed from memory: "
    f"{removed_variables}"
)

print(
    f"Garbage-collected objects: "
    f"{collected_objects}"
)

print(
    f"Score workbook: "
    f"{SCORE_FILE}"
)
```

```python
import gc
import io
import importlib
import os

from contextlib import redirect_stdout

import engines.FiveMinuteCrossSectionalRanker as ranker_module


# =========================================================
# Configuration
# =========================================================

RANKING_FILE = (
    "/home/devinderjeet/fmrss/report/"
    "RankingShares.xlsx"
)


# =========================================================
# Validate Score Data
# =========================================================

if (
    "score_data" not in globals()
    or not score_data
):
    raise RuntimeError(
        "score_data is missing or empty.\n"
        "Run FiveMinuteMathematicalScoreEngine first."
    )


input_symbol_count = len(
    score_data
)


# =========================================================
# Reload Ranker
# =========================================================

importlib.reload(
    ranker_module
)


FiveMinuteCrossSectionalRanker = (
    ranker_module.FiveMinuteCrossSectionalRanker
)


# =========================================================
# Create Ranker
# =========================================================

ranker = FiveMinuteCrossSectionalRanker(
    maximum_selected_symbols=5,

    mathematical_weight=0.60,
    residual_weight=0.25,
    volume_weight=0.15,

    ranking_output_file=RANKING_FILE,

    save_excel=True,
    memory_optimized=True,
)


# =========================================================
# Rank Complete Score Data Silently
# =========================================================

suppressed_output = io.StringIO()


with redirect_stdout(
    suppressed_output
):

    ranked_data, ranking_report = (
        ranker.rank_all(
            score_data=score_data
        )
    )


suppressed_output.close()


# =========================================================
# Validate Result
# =========================================================

if (
    ranking_report is None
    or ranking_report.empty
):
    raise RuntimeError(
        "Cross-Sectional Ranker returned "
        "an empty ranking report.\n"
        "score_data was retained for debugging."
    )


if not os.path.isfile(RANKING_FILE):
    raise RuntimeError(
        f"Ranking workbook was not created: "
        f"{RANKING_FILE}\n"
        "score_data was retained for debugging."
    )


# =========================================================
# Calculate Summary
# =========================================================

successful_symbols = int(
    ranking_report[
        "Status"
    ]
    .eq("CALCULATED")
    .sum()
)


selected_symbol_count = int(
    ranking_report[
        "Latest_Selected"
    ]
    .fillna(False)
    .sum()
)


accepted_symbol_count = int(
    ranking_report[
        "Latest_Selection_Reason"
    ]
    .isin(
        [
            "SELECTED",
            "OUTSIDE_TOP_N",
        ]
    )
    .sum()
)


rejected_symbol_count = int(
    ranking_report[
        "Latest_Selection_Reason"
    ]
    .eq("SETUP_NOT_ACCEPTED")
    .sum()
)


ranking_workbook = (
    ranking_report.attrs.get(
        "Excel_File"
    )
    or RANKING_FILE
)


# =========================================================
# Clear Mathematical Score Memory
# =========================================================

removed_variables = []


for variable_name in [
    "score_data",
    "score_report",
]:

    if variable_name in globals():

        del globals()[variable_name]

        removed_variables.append(
            variable_name
        )


collected_objects = gc.collect()


# =========================================================
# Display Summary
# =========================================================

print("=" * 120)

print(
    f"Successfully processed : "
    f"{successful_symbols}/"
    f"{input_symbol_count}"
)

print(
    f"Accepted mathematical setups : "
    f"{accepted_symbol_count}"
)

print(
    f"Rejected mathematical setups : "
    f"{rejected_symbol_count}"
)

print(
    f"Latest selected symbols : "
    f"{selected_symbol_count}"
)

print(
    f"Ranking workbook : "
    f"{ranking_workbook}"
)

print(
    f"Removed from memory : "
    f"{removed_variables}"
)

print(
    f"Garbage-collected objects : "
    f"{collected_objects}"
)

print("=" * 120)
```
